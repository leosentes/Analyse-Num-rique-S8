```python
def resoudre_edo (f , y0 , t0 , tf , delta_t, schema):
    N = int( ceil (( tf - t0 )/ delta_t ) )
    delta_t = ( tf - t0 )/N
    tn = np . zeros ( N+1 )
    yn = np . zeros ( N+1 )
# complÃľter ici
    tn[0]=t0
    yn[0]=y0
    for i in range(N):
        tn[i+1] = tn[0] + i*delta_t
        yn[i+1] = schema(f , tn[i] , yn[i] , delta_t)
    return tn , yn
```

```python
def Heun(f , tn , yn , delta_t ):
    def phi(tn, yn, delta_t):
        return 0.5*(f(tn, yn) + f(tn+delta_t, yn + delta_t*f(tn, yn))) 
    return yn +delta_t*phi(tn, yn, delta_t)
    
```

```python
def Runge_Kutta4(f , tn , yn , delta_t ):
    k1 = f(tn, yn)
    k2 = f(tn + delta_t/2, yn + delta_t/2 * k1)
    k3 = f(tn + delta_t/2, yn + delta_t/2 * k2)
    k4 = f(tn + delta_t, yn + delta_t * k3)

    return yn + delta_t/6 * (k1 + 2*k2 + 2*k3 + k4)
   
```

```python
def euler_implicite(f, tn, yn, delta_t):

    tn1 = tn + delta_t
    yn1 = yn 
    
    for _ in range(10):
        y_old = yn1
        # Formule implicite : on utilise la nouvelle estimation à droite
        yn1 = yn + delta_t * f(tn1, yn1)
        
        # Critère de convergence
        if abs(yn1 - y_old) < 1e-6:
            break
            
    return yn1
```

```python
def Crank_nicolson(f, tn, yn, delta_t):

    tn1 = tn + delta_t
    yn1 = yn
    
    while True :
        y_mem = yn1
        # Formule implicite : on utilise la nouvelle estimation à droite
        yn1 = yn + (delta_t/2) * (f(tn, yn)+f(tn1, yn1))
        
        # Critère de convergence, 1e-6 fixé arbitrairement 
        if abs(yn1 - y_mem) < 1e-6:
            break
            
    return yn1
```

```python
import matplotlib.pyplot as plt

# Définition du problème
f = lambda t, y: np.exp(-10 * y)
y0 = 2
t0 = 0
tf = 5
delta_t = 0.1

# Résolution avec chaque méthode
tn_ee, yn_ee = resoudre_edo(f, y0, t0, tf, delta_t, euler_explicite)
tn_heun, yn_heun = resoudre_edo(f, y0, t0, tf, delta_t, Heun)
tn_rk4, yn_rk4 = resoudre_edo(f, y0, t0, tf, delta_t, Runge_Kutta4)
tn_ei, yn_ei = resoudre_edo(f, y0, t0, tf, delta_t, euler_implicite)
tn_cn, yn_cn = resoudre_edo(f, y0, t0, tf, delta_t, Crank_nicolson)

# Tracé
plt.figure(figsize=(10, 6))

plt.figure(figsize=(10, 6))

plt.plot(tn_ee,   yn_ee,   label="Euler explicite",   linestyle="-",  marker="")
plt.plot(tn_heun, yn_heun, label="Heun",              linestyle="--", marker="")
plt.plot(tn_rk4,  yn_rk4,  label="Runge-Kutta 4",     linestyle="-.", marker="")
plt.plot(tn_ei,   yn_ei,   label="Euler implicite",   linestyle=":",  marker="")
plt.plot(tn_cn,   yn_cn,   label="Crank-Nicolson",    linestyle=(0, (3, 1, 1, 1)), marker="")

plt.xlabel("t")
plt.ylabel("y(t)")
plt.title("Résolution numérique de $y' = e^{-10y}$, $y(0) = 2$")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

f_test = lambda t, y: -y
y_exacte = lambda t: np.exp(-t)
y0 = 1
t0 = 0
tf = 5

liste_delta_t = [0.5, 0.2, 0.1, 0.05, 0.02, 0.01, 0.005]

methodes = {
    "Euler explicite" : euler_explicite,
    "Heun"            : Heun,
    "Runge-Kutta 4"   : Runge_Kutta4,
    "Euler implicite" : euler_implicite,
    "Crank-Nicolson"  : Crank_nicolson,
}

# Calcul des erreurs
erreurs = {nom: [] for nom in methodes}

for delta_t in liste_delta_t:
    for nom, schema in methodes.items():
        tn, yn = resoudre_edo(f_test, y0, t0, tf, delta_t, schema)
        y_ex = y_exacte(tn)
        erreur_max = np.max(np.abs(yn - y_ex))
        erreurs[nom].append(erreur_max)

# --- Estimation des ordres par régression ---
print("Estimation numérique des ordres de convergence :\n")
print(f"{'Méthode':<20} {'Ordre estimé'}")
print("-" * 35)

ordres = {}
for nom, err in erreurs.items():
    err = np.array(err)
    dt  = np.array(liste_delta_t)
    p = np.polyfit(np.log(dt), np.log(err), 1)
    ordres[nom] = p[0]
    print(f"{nom:<20} {p[0]:.2f}")

# --- Tracé log-log avec droites de référence ancrées ---
plt.figure(figsize=(10, 6))

dt_ref = np.array(liste_delta_t)

for nom, err in erreurs.items():
    line, = plt.loglog(liste_delta_t, err, marker="o", label=nom)

# Droites de référence ancrées sur le premier point de chaque ordre
ref_orders = [1, 2, 4]
styles = ["--", ":", "-."]
colors = ["black", "gray", "lightgray"]

for ordre, style, color in zip(ref_orders, styles, colors):
    # On ancre la droite sur le premier point de dt
    C = erreurs["Euler explicite"][0] / (liste_delta_t[0] ** ordre) \
        if ordre == 1 else \
        erreurs["Heun"][0] / (liste_delta_t[0] ** ordre) \
        if ordre == 2 else \
        erreurs["Runge-Kutta 4"][0] / (liste_delta_t[0] ** ordre)
   
    plt.loglog(dt_ref, C * dt_ref**ordre,
               linestyle=style, color=color,
               label=f"Ordre {ordre} (réf.)")

plt.xlabel("$\\Delta t$")
plt.ylabel("Erreur max $\\|y_n - y(t_n)\\|_\\infty$")
plt.title("Convergence des méthodes numériques (échelle log-log)")
plt.legend()
plt.grid(True, which="both")
plt.tight_layout()
plt.show()
```
